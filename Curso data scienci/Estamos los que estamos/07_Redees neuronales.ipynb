```python
#Importamos librerias necesarias para el análisis
import pandas as pd
import numpy as np
import tensorflow as tf
from tensorflow import keras
from tensorflow.keras import layers
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import LabelEncoder, StandardScaler
import matplotlib.pyplot as plt
```

```python
#Trabajar con el dataset
data = pd.read_csv('titanic.csv')
```

```python
data.head()
```

```python
#2.Pre procesar los datos
#Seleccionar las caracteristicas reelevantes
features= ['Pclass', 'Sex', 'Age', 'SibSp', 'Parch', 'Fare', 'Embarked']
data = data[features + ['Survived']].dropna()
```

```python
data
```

```python
label_encoders= {}
for column in ['Sex','Embarked']:
  le= LabelEncoder()
  data[column]= le.fit_transform(data[column])
  label_encoders[column]=le

label_encoders
```

```python
#Separar características (x) y etiquetas (y)
X = data.drop('Survived', axis=1).values
y = data['Survived'].values
```

```python
X
```

```python
y
```

```python
#Escalar el valor de las variables
scaler=StandardScaler()
X_scaled=scaler.fit_transform(X)
```

```python
#Dvidir el conjunto de datos en entrenamiento 80% y prueba 20%
X_train, X_test, y_train, y_test = train_test_split(X_scaled, y, test_size=0.2, random_state=42)
```

```python
model= keras.Sequential([
    layers.Dense(16, activation='relu', input_shape=(7,)),
    layers.Dense(8, activation='relu'),
    layers.Dense(1, activation='sigmoid')])
```

```python
model.compile(optimizer='adam', loss='binary_crossentropy', metrics=['accuracy'])
```

```python
model.summary()
```

```python
history=model.fit(X_train, y_train, epochs=300, batch_size=32, validation_split=0.2, verbose=1)
```

```python
# 8.- Predicciones y Métricas de Evaluación
from sklearn.metrics import accuracy_score, precision_score, recall_score, f1_score, confusion_matrix, ConfusionMatrixDisplay

# Obtener predicciones en probabilidades y convertirlas a clases binarias (0 o 1)
y_pred_prob = model.predict(X_test)
y_pred = (y_pred_prob >= 0.5).astype(int)

# Cálculo de métricas
acc = accuracy_score(y_test, y_pred)
prec = precision_score(y_test, y_pred)
rec = recall_score(y_test, y_pred)
f1 = f1_score(y_test, y_pred)

print("--- Métricas de Rendimiento (Red Neuronal) ---")
print(f"Accuracy:  {acc:.4f}")
print(f"Precision: {prec:.4f}")
print(f"Recall:    {rec:.4f}")
print(f"F1-Score:  {f1:.4f}")

# Matriz de confusión
cm = confusion_matrix(y_test, y_pred)
print("\nMatriz de Confusión:")
print(cm)

# Graficar la matriz de confusión
disp = ConfusionMatrixDisplay(confusion_matrix=cm, display_labels=['No Sobrevivió', 'Sobrevivió'])
disp.plot(cmap=plt.cm.Blues)
plt.title("Matriz de Confusión - Red Neuronal")
plt.show()
```
